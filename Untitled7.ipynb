```python
²ŵimport math

class KDNode:
    """
    Represents a single node in the 2-D kD Tree.
    Each node stores a spatial point and pointers to its left and right children.
    """
    def __init__(self, point, left=None, right=None):
        self.point = point  # Tuple format: (sensor_id, x, y)
        self.left = left    # Left child KDNode
        self.right = right  # Right child KDNode

class ForestFireMonitoringSystem:
    """
    A dynamic spatial monitoring system using a 2-D kD Tree to track
    IoT fire sensors across a wildlife reserve.
    """
    def __init__(self):
        # Master list to keep track of all active sensors and validate duplicates
        self.sensor_list = []
        # Root node of the kD Tree
        self.root = None
        # Tracks the number of nodes visited during the most recent spatial query
        self.last_nodes_visited = 0

    def calculate_height(self, node):
        """Recursively calculates the maximum depth/height of the tree."""
        if not node:
            return 0
        return 1 + max(self.calculate_height(node.left), self.calculate_height(node.right))

    def _insert_node(self, node, point, depth):
        """
        Dynamically inserts a new node as a leaf without fully rebuilding the tree.
        Alternates between X and Y axes based on the current depth.
        """
        # Base case: Found an empty spot, insert the new node here
        if node is None:
            return KDNode(point)

        # Determine the current splitting axis (0 for X-axis, 1 for Y-axis)
        axis = depth % 2

        # Compare the point's coordinate on the active axis with the current node's coordinate
        # point[1] is X, point[2] is Y. Hence point[axis + 1] gets the correct coordinate.
        if point[axis + 1] < node.point[axis + 1]:
            node.left = self._insert_node(node.left, point, depth + 1)
        else:
            node.right = self._insert_node(node.right, point, depth + 1)

        return node

    def add_sensor(self, sid, x, y, outfile):
        """Adds a new sensor to the network and the spatial tree."""
        try:
            fx, fy = float(x), float(y)

            # 1. Linear scan to ensure the sensor ID is unique
            for sensor in self.sensor_list:
                if sensor[0] == sid:
                    outfile.write(f"Error Adding Sensor: Sensor {sid} already exists.\n")
                    return

            point = (sid, fx, fy)
            # 2. Append to tracking list
            self.sensor_list.append(point)
            # 3. Dynamically insert into the KD-Tree
            self.root = self._insert_node(self.root, point, 0)

            outfile.write(f"Sensor Added:\n{sid} ({int(fx)},{int(fy)})\n")
        except ValueError:
            outfile.write(f"Error Adding Sensor: Invalid coordinates received for {sid}.\n")
        except Exception as e:
            outfile.write(f"Error Adding Sensor: {str(e)}\n")

    def find_min(self, node, axis, depth):
        """Finds the node with the minimum value along a specific target axis."""
        if node is None:
            return None

        curr_axis = depth % 2

        # If the current splitting axis matches our target axis,
        # the minimum MUST be in the left subtree.
        if curr_axis == axis:
            if node.left is None:
                return node
            return self.find_min(node.left, axis, depth + 1)

        # Otherwise, the minimum could be in the left subtree, right subtree, or the current node
        left_min = self.find_min(node.left, axis, depth + 1)
        right_min = self.find_min(node.right, axis, depth + 1)

        res = node
        if left_min and left_min.point[axis + 1] < res.point[axis + 1]:
            res = left_min
        if right_min and right_min.point[axis + 1] < res.point[axis + 1]:
            res = right_min
        return res

    def find_max(self, node, axis, depth):
        """Finds the node with the maximum value along a specific target axis."""
        if node is None:
            return None

        curr_axis = depth % 2

        # If the current splitting axis matches our target axis,
        # the maximum MUST be in the right subtree.
        if curr_axis == axis:
            if node.right is None:
                return node
            return self.find_max(node.right, axis, depth + 1)

        # Otherwise, check both subtrees and the current node
        left_max = self.find_max(node.left, axis, depth + 1)
        right_max = self.find_max(node.right, axis, depth + 1)

        res = node
        if left_max and left_max.point[axis + 1] > res.point[axis + 1]:
            res = left_max
        if right_max and right_max.point[axis + 1] > res.point[axis + 1]:
            res = right_max
        return res

    def _delete_node(self, node, target_point, depth):
        """
        Deletes a node from the KD-Tree using structural replacement.
        Replaces deleted nodes with the maximum of the left subtree or minimum of the right subtree.
        """
        if node is None:
            return None

        axis = depth % 2

        # Case 1: The current node is the target node to be deleted
        if node.point == target_point:
            # If left child exists, find the max in left subtree, replace, and delete that max node
            if node.left is not None:
                rep = self.find_max(node.left, axis, depth + 1)
                node.point = rep.point
                node.left = self._delete_node(node.left, rep.point, depth + 1)
            # If right child exists, find the min in right subtree, replace, and delete that min node
            elif node.right is not None:
                rep = self.find_min(node.right, axis, depth + 1)
                node.point = rep.point
                node.right = self._delete_node(node.right, rep.point, depth + 1)
            # If it's a leaf node, simply remove it
            else:
                return None

        # Case 2: Target is smaller on current axis, traverse left
        elif target_point[axis + 1] < node.point[axis + 1]:
            node.left = self._delete_node(node.left, target_point, depth + 1)

        # Case 3: Target is greater or equal on current axis, traverse right
        else:
            node.right = self._delete_node(node.right, target_point, depth + 1)

        return node

    def remove_sensor(self, sid, outfile):
        """Removes a sensor from the system based on its ID."""
        try:
            target_point = None
            # 1. Locate the sensor in the tracking list
            for i, sensor in enumerate(self.sensor_list):
                if sensor[0] == sid:
                    target_point = sensor
                    self.sensor_list.pop(i)
                    break

            # 2. If found, structurally delete it from the KD-Tree
            if target_point:
                self.root = self._delete_node(self.root, target_point, 0)
                outfile.write(f"Sensor Removed:\n{sid}\n")
            else:
                outfile.write(f"Error Removing Sensor: Sensor {sid} not found.\n")
        except Exception as e:
            outfile.write(f"Error Removing Sensor: {str(e)}\n")

    def update_sensor(self, sid, new_x, new_y, outfile):
        """Updates the coordinates of an existing sensor."""
        try:
            fx, fy = float(new_x), float(new_y)
            target_point = None

            # 1. Find the sensor to update
            for i, sensor in enumerate(self.sensor_list):
                if sensor[0] == sid:
                    target_point = sensor
                    self.sensor_list[i] = (sid, fx, fy)
                    break

            # 2. To update, we must delete the old coordinate node and insert the new one
            if target_point:
                self.root = self._delete_node(self.root, target_point, 0)
                self.root = self._insert_node(self.root, (sid, fx, fy), 0)
                outfile.write(f"Sensor Updated:\n{sid} ({int(fx)},{int(fy)})\n")
            else:
                outfile.write(f"Error Updating Sensor: Sensor {sid} not found.\n")
        except ValueError:
            outfile.write(f"Error Updating Sensor: Invalid coordinates received for {sid}.\n")
        except Exception as e:
            outfile.write(f"Error Updating Sensor: {str(e)}\n")

    def find_nearest(self, qx, qy, outfile):
        """Finds the spatially nearest sensor to a given query coordinate using hypersphere pruning."""
        try:
            qx, qy = float(qx), float(qy)
            # best stores [best_node_point, best_distance]
            best = [None, float('inf')]
            nodes_visited = [0]

            def search(node, depth):
                if node is None:
                    return

                sid, sx, sy = node.point

                # Calculate Euclidean distance
                dist = math.hypot(sx - qx, sy - qy)

                # Increment nodes visited only when the best distance is successfully updated
                if dist < best[1]:
                    best[0] = node.point
                    best[1] = dist
                    nodes_visited[0] += 1

                axis = depth % 2
                q_val = qx if axis == 0 else qy
                n_val = sx if axis == 0 else sy

                # Determine which branch contains the query point (near side)
                next_b = node.left if q_val < n_val else node.right
                other_b = node.right if q_val < n_val else node.left

                # 1. Traverse down the near side first
                search(next_b, depth + 1)

                # 2. Only search the far side if the hypersphere crosses the splitting plane
                if abs(q_val - n_val) < best[1]:
                    search(other_b, depth + 1)

            search(self.root, 0)
            self.last_nodes_visited = nodes_visited[0]

            if best[0]:
                outfile.write(f"Nearest Sensor\n{best[0][0]} ({int(best[0][1])},{int(best[0][2])})\nDistance: {best[1]:.2f}\nNodes Visited: {nodes_visited[0]}\n")
            else:
                outfile.write("Nearest Sensor Search: No sensors found in the tree.\n")
        except ValueError:
            outfile.write("Error Nearest Sensor: Invalid query coordinates provided.\n")
        except Exception as e:
            outfile.write(f"Error Nearest Sensor: {str(e)}\n")

    def find_in_rectangle(self, xmin, ymin, xmax, ymax, outfile):
        """Finds all sensors inside a bounding box using axis-aligned pruning."""
        try:
            xmin, ymin, xmax, ymax = float(xmin), float(ymin), float(xmax), float(ymax)

            # Auto-correction in case bounding box dimensions are passed inverted
            if xmin > xmax: xmin, xmax = xmax, xmin
            if ymin > ymax: ymin, ymax = ymax, ymin

            results = []
            nodes_visited = [0]

            def search(node, depth):
                if node is None:
                    return

                sid, sx, sy = node.point

                # Increment nodes visited only when successfully adding a node inside the region
                if xmin <= sx <= xmax and ymin <= sy <= ymax:
                    results.append(node.point)
                    nodes_visited[0] += 1

                axis = depth % 2
                n_val = sx if axis == 0 else sy

                # Spatial Pruning:
                # If the current region intersects the bounding box, recurse into that side.
                if axis == 0:  # X-axis check
                    if xmin <= n_val:
                        search(node.left, depth + 1)
                    if xmax >= n_val:
                        search(node.right, depth + 1)
                else:          # Y-axis check
                    if ymin <= n_val:
                        search(node.left, depth + 1)
                    if ymax >= n_val:
                        search(node.right, depth + 1)

            search(self.root, 0)
            self.last_nodes_visited = nodes_visited[0]

            outfile.write(f"Sensors inside Region\n({int(xmin)},{int(ymin)}) to ({int(xmax)},{int(ymax)})\n")
            for r in results:
                outfile.write(f"{r[0]} ({int(r[1])},{int(r[2])})\n")
            outfile.write(f"Total Sensors: {len(results)}\n")
        except ValueError:
            outfile.write("Error Range Search: Invalid region coordinates provided.\n")
        except Exception as e:
            outfile.write(f"Error Range Search: {str(e)}\n")

    def display_statistics(self, outfile):
        """Displays current structural statistics of the KD Tree."""
        try:
            height = self.calculate_height(self.root)
            outfile.write(f"Tree Statistics\nTotal Sensors : {len(self.sensor_list)}\nTree Height :{height}\nNodes Visited (Last Query): {self.last_nodes_visited}\n")
        except Exception as e:
            outfile.write(f"Error Displaying Statistics: {str(e)}\n")

    def display_traversal(self, outfile):
        """Generates a Preorder sequence traversal (Root -> Left -> Right)."""
        try:
            outfile.write("Preorder Traversal\n")
            def preorder(node):
                if node:
                    outfile.write(f"{node.point[0]}\n")
                    preorder(node.left)
                    preorder(node.right)
            preorder(self.root)
        except Exception as e:
            outfile.write(f"Error Displaying Traversal: {str(e)}\n")

    def process_file(self, input_filename, output_filename):
        """Processes commands dynamically from the input text file."""
        with open(input_filename, 'r') as infile, open(output_filename, 'w') as outfile:
            for line in infile:
                line = line.strip()
                # Skip blank lines and comments
                if not line or line.startswith("#"):
                    continue

                parts = line.split("::")
                cmd = parts[0].strip()

                if cmd == "addSensor":
                    self.add_sensor(parts[1], parts[2], parts[3], outfile)
                elif cmd == "removeSensor":
                    self.remove_sensor(parts[1], outfile)
                elif cmd == "nearestSensor":
                    self.find_nearest(parts[1], parts[2], outfile)
                elif cmd == "rangeSearch":
                    self.find_in_rectangle(parts[1], parts[2], parts[3], parts[4], outfile)
                elif cmd == "treeStatistics":
                    self.display_statistics(outfile)
                elif cmd == "displayTree":
                    self.display_traversal(outfile)
                elif cmd == "updateSensor":
                    self.update_sensor(parts[1], parts[2], parts[3], outfile)

if __name__ == "__main__":
    input_file = "inputPS8.txt"
    output_file = "outputPS8.txt"

    system = ForestFireMonitoringSystem()
    system.process_file(input_file, output_file)
```
